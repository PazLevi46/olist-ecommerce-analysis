## Data Model Summary

### Customers
- **Grain:** One row per customer record associated with an order.
- **Primary Key:** `customer_id`
- **Customer Identifier:** `customer_unique_id` identifies the same customer across multiple orders.
- **Connects to:** `orders` via `customer_id`
- **Geographic reference:** `customer_zip_code_prefix` can be linked to `geolocation`.

### Orders
- **Grain:** One row per order.
- **Primary Key:** `order_id`
- **Foreign Key:** `customer_id`
- **Connects to:** `customers`, `order_items`, `order_payments`, and `order_reviews`.

### Order Items
- **Grain:** One row per item within an order.
- **Composite Key:** `order_id` + `order_item_id`
- **Foreign Keys:** `order_id`, `product_id`, `seller_id`
- **Connects to:** `orders`, `products`, and `sellers`.

### Products
- **Grain:** One row per product.
- **Primary Key:** `product_id`
- **Category reference:** `product_category_name`
- **Connects to:** `order_items` via `product_id` and `product_category_translation` via `product_category_name`.

### Product Category Translation
- **Grain:** One row per product category.
- **Primary Key:** `product_category_name`
- **Connects to:** `products` via `product_category_name`.

### Sellers
- **Grain:** One row per seller.
- **Primary Key:** `seller_id`
- **Connects to:** `order_items` via `seller_id`.
- **Geographic reference:** `seller_zip_code_prefix` can be linked to `geolocation`.

### Order Payments
- **Grain:** One payment component within an order.
- **Composite Key:** `order_id` + `payment_sequential`
- **Foreign Key:** `order_id`
- **Connects to:** `orders`.

### Order Reviews
- **Grain:** One review–order association.
- **Composite Key:** `review_id` + `order_id`
- **Foreign Key:** `order_id`
- **Connects to:** `orders`.
- **Important note:** Multiple reviews may be associated with the same order, and the same review record may appear across multiple orders. This will require further handling before order-level analysis.

### Geolocation
- **Grain:** One geolocation record associated with a ZIP code prefix.
- **Primary Key:** No natural primary key identified in the raw dataset.
- **Reference Key:** `geolocation_zip_code_prefix`
- **Connects to:** `customers` and `sellers` through their respective ZIP code prefix fields.

#### Carrier delivery before purchase

```python
## Inconsistency between the time of the delivery to the carrier and the purchase time
carrier_before_purchase_rows = orders[
    orders['order_delivered_carrier_date'] <
    orders['order_purchase_timestamp']
].copy()

# Finding the time difference
carrier_before_purchase_rows['time_difference'] = (
    carrier_before_purchase_rows['order_purchase_timestamp'] -
    carrier_before_purchase_rows['order_delivered_carrier_date']
)

# Time difference distribution
carrier_before_purchase_rows['time_difference'].describe()

# Dividing the time differences to 3 groups to better understand the distributions
print(
    "Up to 1 hour:",
    (carrier_before_purchase_rows['time_difference'] <= pd.Timedelta(hours=1)).sum()
)

print(
    "1–24 hours:",
    (
        (carrier_before_purchase_rows['time_difference'] > pd.Timedelta(hours=1)) &
        (carrier_before_purchase_rows['time_difference'] <= pd.Timedelta(days=1))
    ).sum()
)

print(
    "More than 1 day:",
    (carrier_before_purchase_rows['time_difference'] > pd.Timedelta(days=1)).sum()
)

# Looking into the records with the largest time difference
carrier_before_purchase_rows.sort_values(
    'time_difference',
    ascending=False
)[[
    'order_id',
    'order_purchase_timestamp',
    'order_delivered_carrier_date',
    'time_difference'
]].head(10)
```

Customer-before-Carrier inconsistencies had a median difference of approximately 1 day and 16 hours, while Customer-before-Approval inconsistencies had a median difference of approximately 1 day.

#### Timeline inconsistencies

```python
## Merging the order_item table and the orders table on the order)id column
order_items_check = order_items.merge(
    orders[['order_id', 'order_purchase_timestamp', 'order_approved_at']],
    on='order_id',
    how='left'
)

## Summing the number of occasions where the shipping limit date was either before the purchase or before the approval
print(
    "Shipping deadline before purchase:",
    (order_items_check['shipping_limit_date'] <
     order_items_check['order_purchase_timestamp']).sum()
)

print(
    "Shipping deadline before approval:",
    (order_items_check['shipping_limit_date'] <
     order_items_check['order_approved_at']).sum()
)

## There are 127 records with shipping deadline before approval date
## Chacking the time difference between them
deadline_before_approval = order_items_check[
    order_items_check['shipping_limit_date'] <
    order_items_check['order_approved_at']
].copy()

deadline_before_approval['time_difference'] = (
    deadline_before_approval['order_approved_at'] -
    deadline_before_approval['shipping_limit_date']
)

## The time differernce statistics
deadline_before_approval['time_difference'].describe()

## Dividing into 3 categories of time differences
print(
    "Up to 1 hour:",
    (deadline_before_approval['time_difference'] <= pd.Timedelta(hours=1)).sum()
)

print(
    "1–24 hours:",
    (
        (deadline_before_approval['time_difference'] > pd.Timedelta(hours=1)) &
        (deadline_before_approval['time_difference'] <= pd.Timedelta(days=1))
    ).sum()
)

print(
    "More than 1 day:",
    (deadline_before_approval['time_difference'] > pd.Timedelta(days=1)).sum()
)

## Looking at the values with the time difference in the descending order
deadline_before_approval[
    ['order_id', 'shipping_limit_date', 'order_approved_at', 'time_difference']
].sort_values('time_difference', ascending=False)
```

#### Shipping Limit Date Validation

The `shipping_limit_date` was compared with the corresponding purchase and approval timestamps from the `orders` table.

- No order items had a shipping deadline earlier than the order purchase timestamp.
- 127 order items had a shipping deadline earlier than the recorded order approval timestamp.
- Of these 127 records:
  - 7 had a difference of up to 1 hour.
  - 60 had a difference between 1 and 24 hours.
  - 60 had a difference greater than 24 hours.
- The largest observed difference was approximately 18.5 days.
- Several of the larger inconsistencies occurred around similar dates, suggesting a recurring pattern rather than an isolated record.

The source of these chronological inconsistencies cannot be determined from the available data. Therefore, no timestamps were modified and no records were removed. These records should be treated with caution in analyses that depend specifically on the relationship between order approval and the seller shipping deadline.

## Final Data Cleaning Summary

The nine datasets were reviewed for data types, missing values, duplicate records, categorical consistency, numerical validity, and relevant chronological inconsistencies. Relationships between tables were also considered where necessary to evaluate the impact of data quality issues.

### Data Type and Formatting Adjustments

- Date and timestamp columns were converted from `object` to `datetime` where required.
- Identifier fields that do not represent numerical quantities, such as ZIP code prefixes, were treated as strings.
- Product name length, description length, and photo quantity were converted to nullable integer (`Int64`) types.
- Leading and trailing whitespace was removed from relevant categorical text fields.

### Missing Values

Missing values were evaluated based on their business meaning rather than being automatically removed or imputed.

- Missing order timestamps were largely consistent with the corresponding order status and therefore represent expected missingness for orders that had not reached those stages.
- A small number of delivered orders were missing timestamps from earlier stages of the order lifecycle. These records were retained because the missing values could not be reliably inferred.
- 610 products are missing category, name-length, description-length, and photo-quantity information. All 610 products appear in `order_items`, accounting for 1603 order-item records, so they were retained.
- Two of these products are also missing physical dimensions and weight.
- Missing review titles and comments were retained, since customers can provide a numerical review score without written feedback.

No missing values were imputed where the true value could not be reliably inferred.

### Numerical and Categorical Validation

- Product prices, freight values, payment values, product dimensions, and other numerical attributes were reviewed for unusual or invalid values.
- Zero freight values were retained because zero-cost shipping is a plausible business scenario.
- Zero-value payment records were investigated and retained based on their payment and order context.
- Two credit-card payment records contain zero recorded installments despite positive payment values. Since the correct installment values cannot be inferred, they were retained.
- Four products have a recorded weight of 0 grams despite having non-zero dimensions. These values were retained because their correct weights could not be reliably inferred.
- Review scores contained only the expected values from 1 to 5.
- Geographic coordinates were within valid latitude and longitude ranges.
- Relevant categorical fields were checked for formatting and consistency issues.

### Chronological Validation

Order lifecycle timestamps were reviewed for chronological inconsistencies.

Several inconsistencies were identified between recorded order stages, including cases where carrier or delivery timestamps occurred before earlier recorded stages. The magnitude of these differences was investigated rather than assuming that all inconsistencies represented data-entry errors.

Additionally, 127 order-item records have a seller shipping deadline earlier than the recorded order approval timestamp.

Since the available data does not provide enough information to determine the source or correct value of these timestamp inconsistencies, the records were retained without modification. They should be treated carefully in analyses that depend on the affected timestamp relationships.

### Duplicate and Key Validation

Primary and composite keys were evaluated during the initial data-understanding stage.

- No unnecessary full-row duplicates requiring removal were identified during cleaning.
- The reviews table contains non-unique relationships between `review_id` and `order_id`: 789 review IDs and 547 order IDs appear more than once.
- Inspection showed that repeated order IDs may correspond to different review records, while repeated review IDs may also be associated with different orders.
- Since there is insufficient information to determine which records, if any, are incorrect, these records were retained.

The non-unique relationship between orders and reviews must be considered when calculating order-level review metrics.

### Final Cleaning Decisions

The cleaning process prioritized preserving valid business information rather than removing records solely because they contained incomplete or unusual values.

No values were artificially imputed when the correct value could not be determined, and records with partial data were retained when their remaining fields were still analytically useful.

Known data-quality limitations were documented and will be handled according to the requirements of each specific analysis.
